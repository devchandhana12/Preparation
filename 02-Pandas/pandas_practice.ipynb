```python
import pandas as pd

df = pd.read_csv("Salary_dataset.csv")
salary_df = df[["YearsExperience","Salary"]].head(5)
salary_df
```

```python
experienced_employees = df["YearsExperience"] >=5.0
df[experienced_employees].head(5)
```

```python
# sorting based on salary
sorted_salary = df.sort_values(by="Salary", ascending=False)
sorted_salary.head(5)

# Monthly salary
df["MonthlySalary"] = df["Salary"] / 12
df[["Salary", "MonthlySalary"]].head(5)


# Summarizing stuff
avg_salary_by_experience = df.groupby(experienced_employees)["Salary"].mean()
avg_salary_by_experience
df["IsExperienced"] = experienced_employees

df[["YearsExperience", "IsExperienced"]].head()

salary_summary = df.groupby(experienced_employees)["Salary"].agg([
    "min", "max","mean"
])

missing_salary = df["Salary"].isna()
missing_salary.sum()
```

```python
df = df.rename(columns={"Unnamed: 0": "ID"})

duplicate_ids = df.duplicated(subset=["ID"], keep=False)
df[duplicate_ids].sum()

df["SalaryRounded"] = df["Salary"].round().astype("int64")
df.head(5)

salary_columns = df[["Salary", "SalaryRounded"]]




experienced_pay = df.loc[
    df["YearsExperience"] >=5,
    ["YearsExperience","MonthlySalary"]
]
experienced_pay.head(5)

# iloc works with indexes
experienced_pay.iloc[5:10, [0,1]]
```

```python
filtered_employees = df.loc[
    (df["YearsExperience"] >= 5) & (df["MonthlySalary"] > 7000)
 ]

filtered_employees.head(5)
```

```python
either_condition = df.loc[
    (df["YearsExperience"] < 2 ) | (df["MonthlySalary"] > 7000)
]

either_condition.head(5)
```

```python
experience_labels = pd.DataFrame({
    "IsExperienced": [False, True],
    "ExperienceGroup": ["Under 5 years", "5 years or more"],
})

labeled_salaries = df.merge(
    experience_labels,
    # "on" should exist in both tables
    on="IsExperienced",
    how="left"
)

labeled_salaries[["YearsExperience","ExperienceGroup"]].head(5)


```

```python
experience_labels = experience_labels.loc[
    experience_labels["IsExperienced"] == True
]

labeled_salaries = df.merge(
    experience_labels,
    on="IsExperienced",
    how="inner"
)

labeled_salaries.head(5)


experienced = df.loc[df["IsExperienced"] == True]
less_experienced = df.loc[df["IsExperienced"] == False]

combined_list = pd.concat(
    [experienced, less_experienced],
    ignore_index=True
)
combined_list
```

```python
df["BonusRate"]  = 0.0

df.loc[df["IsExperienced"]== True, "BonusRate"] = 0.05

df.head(5)
```

```python
df["AnnualBonus"] = df["Salary"] * df["BonusRate"]

has_bonus = df["AnnualBonus"] > 0
filtered_employees = df.loc[has_bonus]

ranked_employees = filtered_employees.sort_values(
    by="AnnualBonus",
    ascending=False,
)

ranked_employees[["YearsExperience", "Salary", "AnnualBonus"]].head(5)
```

```python
practice_pay = df[["YearsExperience", "Salary"]].copy()
practice_pay.loc[practice_pay.index[0], "Salary"] = float("nan")

practice_pay["Salary"] = practice_pay["Salary"].fillna(practice_pay["Salary"].median())
practice_pay.isna().sum()
```

```python
messy_labels = pd.DataFrame({
    "ExperienceGroup": [" Under 5 years", "5 years or more  "]
})

messy_labels["ExperienceGroup"] = messy_labels["ExperienceGroup"].str.strip()
messy_labels["ExperienceGroup"].tolist()
```
