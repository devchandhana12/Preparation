```python
# List comprehension
squares = []
for x in range(20):
    if x % 2 == 0:
        squares.append(x ** 2)


squares = [x**2 for x in range(20) if x % 2 == 0]
print(squares)
```

```python
def move_zero(nums:list[int]):
    write = 0
    for i in range(len(nums)):
        if nums[i] != 0:
            nums[write], nums[i] = nums[i], nums[write]
            write+=1
    return nums
nums = [0,1,0,3,12]
print(move_zero(nums))
```

```python
def flaten(nested):
    result = []
    for item in nested:
        if isinstance(item, list):
            result.extend(flaten(item))
        else:
            result.append(item)
    return result

print(flaten([[1,2, [2.00, 2.01]], [3,4]]))
```

```python
def rotate(nums:list, k:int) -> list:
    if not nums:
        return nums
    k = k % len(nums)
    return nums[-k:] + nums[:-k]

print(rotate([1,2,3,4,5],2))
print(rotate([1,2,3],4))
```

```python
def second_largest(nums: list):
    largest = None
    second = None
    for n in nums:
        if largest is None or n > largest:
            second = largest      # paatha largest ippudu second avthundi
            largest = n
        elif n != largest and (second is None or n > second):
            second = n
    return second

print(second_largest([10, 5, 8, 20, 15]))  # 15
print(second_largest([20, 10, 15, 5]))     # 15
print(second_largest([4, 4, 2]))           # 2
print(second_largest([7]))                 # None
```

```python
def missing_number(nums: list) -> int:
    n = len(nums) + 1
    for i in range(1, n + 1):
        if i not in nums:
            return i

print(missing_number([1, 2, 4, 5]))   # 3
```

```python
# snake water gun game
import random

# List of valid choices
choices = ["snake", "water", "gun"]

# Computer makes a random choice
computer_choice = random.choice(choices)

# User makes a choice
user_choice = input("Enter your choice (snake, water, or gun): ").lower()

# Validate user input
if user_choice not in choices:
    print("Invalid choice! Please choose snake, water, or gun.")
else:
    print(f"Computer chose: {computer_choice}")
    print(f"You chose: {user_choice}")

    # Check for a tie
    if computer_choice == user_choice:
        print("It's a tie!")

    # Check all winning conditions for the user
    elif (user_choice == "snake" and computer_choice == "water") or \
         (user_choice == "water" and computer_choice == "gun") or \
         (user_choice == "gun" and computer_choice == "snake"):
        print("You win!")

    # If it's not a tie and the user didn't win, the computer wins
    else:
        print("You lose!")
```

```python
import random

def guessing_game():
    target_num = random.randint(1, 10)

    guesses = 0
    user_guess = None

    print("Welcome to the Number Guessing Game!")
    print("I have picked a number between 1 and 100. Try to guess it!\n")

    while user_guess != target_num:
        try:
            user_guess = int(input("Enter your guess:"))
            guesses +=1

            if user_guess > target_num:
                print("Lower number please \n")
            elif user_guess < target_num:
                print("Higher number please \n")
            else:
                print(f"Congratulations! you guessed the number in {guesses}")
        except ValueError:
            print("Please nothing!")

guessing_game()
```
